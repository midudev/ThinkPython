## ¿Qué es un notebook?

Un Jupyter notebook está compuesto por celdas; cada celda contiene texto o código.
Esta celda contiene texto. 

La siguiente celda contiene código.

```python
print('Hello')
```

Haz clic en la celda anterior para seleccionarla.
Deberías ver un botón a la izquierda con un triángulo dentro de un círculo, que es el icono de "Play".
Si pulsas este botón, Jupyter ejecuta el código de la celda y muestra el resultado.

Cuando ejecutas código en un notebook por primera vez, puede tardar unos segundos en iniciar.
Y si es un notebook que no escribiste tú, puede que aparezca un mensaje de advertencia.
Si estás ejecutando un notebook de una fuente en la que confías, que espero que me incluya, puedes pulsar "Run Anyway".

En lugar de hacer clic en el botón "Play", también puedes ejecutar el código de una celda manteniendo pulsado `Shift` y presionando `Enter`.

Si estás ejecutando este notebook en Colab, deberías ver botones en la parte superior izquierda que dicen "+ Código" y "+ Text". El primero añade una celda de código y el segundo añade una celda de texto.
Si quieres probarlos, selecciona esta celda haciendo clic en ella y luego pulsa el botón "+ Text".
Debería aparecer una nueva celda debajo de esta.

Añade algo de texto a la celda.
Puedes usar los botones para darle formato, o puedes marcar el texto usando [Markdown](https://www.markdownguide.org/basic-syntax/).
Cuando termines, mantén pulsado `Shift` y presiona `Enter`, lo que dará formato al texto que acabas de escribir y pasará a la siguiente celda.

En cualquier momento Jupyter está en uno de dos modos:

* En **command mode**, puedes realizar operaciones que afectan a las celdas, como añadir y eliminar celdas completas.

* En **edit mode**, puedes editar el contenido de una celda.

Con las celdas de texto, es evidente en qué modo estás.
En edit mode, la celda se divide verticalmente, con el texto que estás editando a la izquierda y el texto formateado a la derecha.
Y verás herramientas de edición de texto en la parte superior.
En command mode, solo ves el texto formateado.

Con las celdas de código, la diferencia es más sutil, pero si hay un cursor en la celda, estás en edit mode.

Para pasar de edit mode a command mode, pulsa `ESC`.
Para pasar de command mode a edit mode, pulsa `Enter`.

Cuando termines de trabajar en un notebook, puedes cerrar la ventana, pero cualquier cambio que hayas hecho desaparecerá.
Si haces cambios que quieres conservar, abre el menú File en la parte superior izquierda.
Verás varias formas de guardar el notebook.

* Si tienes una cuenta de Google, puedes guardar el notebook en tu Drive.

* Si tienes una cuenta de GitHub, puedes guardarlo en GitHub.

* O, si quieres guardar el notebook en tu ordenador, selecciona "Download" y luego "Download .ipynb". El sufijo ".ipynb" indica que es un archivo notebook, a diferencia de ".py", que indica un archivo que contiene solo código de Python.

## Código para *Think Python*

Al principio de cada notebook, verás una celda con código como esta:

```python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/thinkpython.py')

import thinkpython
```

No necesitas saber cómo funciona este código, pero cuando llegues al final del libro, la mayor parte tendrá sentido.
Como quizá imagines, descarga un archivo; en concreto, descarga `thinkpython.py`, que contiene código de Python proporcionado específicamente para este libro.
La última línea "importa" este código, lo que significa que podemos usar el código en el notebook.

En otros capítulos, verás código que descarga `diagram.py`, que se usa para generar los diagramas del libro, y `jupyturtle.py`, que se usa en varios capítulos para crear turtle graphics.

En algunos lugares verás una celda como esta que empieza con `%%expect`.

```python
%%expect SyntaxError

abs 42
```

`%%expect` no forma parte de Python: es un "magic command" de Jupyter que indica que esperamos que la celda produzca un error.
Cuando veas este comando, significa que el error es deliberado, normalmente pensado para advertirte sobre una trampa común.
